```python
# Start ollama serve in the background using `nohup`.
# `nohup` makes the command immune to hangups, and `&` runs it in the background.
# Output is redirected to `ollama_serve.log` to keep the console clean.
!nohup ollama serve > ollama_serve.log 2>&1 &
print('Ollama server attempting to start in the background...')
```

```python
import time
import requests

# Give Ollama some initial time to boot up
time.sleep(15)

server_up = False
max_attempts = 10
for i in range(max_attempts):
    try:
        # Attempt to connect to the Ollama API. Use a short timeout for each attempt.
        response = requests.get('http://127.0.0.1:11434/api/tags', timeout=5)
        if response.status_code == 200:
            print("Ollama server is up and reachable!")
            server_up = True
            break
    except requests.exceptions.ConnectionError:
        print(f"Attempt {i+1}/{max_attempts}: Ollama server not yet reachable. Retrying in 5 seconds...")
    except requests.exceptions.Timeout:
        print(f"Attempt {i+1}/{max_attempts}: Connection timed out. Retrying in 5 seconds...")
    time.sleep(5)

if not server_up:
    print("Failed to connect to Ollama server after multiple attempts.")
    print("Checking ollama_serve.log for more details:")
    !cat ollama_serve.log
else:
    print("Successfully connected to Ollama server. Now listing models to confirm.")
    # If connected, list models to confirm operational status
    !curl http://127.0.0.1:11434/api/tags

```

```python
# Ensure Ollama and its dependencies are installed
!apt-get update -qq
!apt-get install -y zstd > /dev/null
!curl -fsSL https://ollama.com/install.sh | sh > ollama_install_log.txt 2>&1
print('Ollama installation process completed. Checking logs if needed:')
!cat ollama_install_log.txt

import os
import time
import requests

# Add /usr/local/bin to PATH for Python subprocess calls, though not directly used for shell commands below
if '/usr/local/bin' not in os.environ['PATH']:
    os.environ['PATH'] = f"{os.environ['PATH']}:/usr/local/bin"

ollama_path = '/usr/local/bin/ollama'

# 1. Kill any existing ollama processes to ensure a clean start.
print('Ensuring no old Ollama processes are running...')
!pkill ollama || true
time.sleep(2) # Give a moment for processes to terminate

# 2. Start ollama serve in the background using `nohup`.
# First, verify the executable itself for better diagnostics.
print(f"Verifying Ollama executable at {ollama_path}...")
if not os.path.exists(ollama_path):
    print(f"ERROR: Ollama executable NOT FOUND at {ollama_path}.")
    print("Please ensure Ollama is installed correctly.")
    # Exit or raise error, for now just print and proceed to fail
elif not os.access(ollama_path, os.X_OK):
    print(f"ERROR: Ollama executable at {ollama_path} is NOT EXECUTABLE.")
    print("Attempting to make it executable...")
    !chmod +x {ollama_path}
    if not os.access(ollama_path, os.X_OK):
        print(f"ERROR: Failed to make {ollama_path} executable. Please check permissions manually.")
else:
    print(f"Ollama executable at {ollama_path} exists and is executable.")

print('Attempting to start Ollama server in the background using bash -c nohup...')
# Use bash -c to ensure the nohup command runs in a proper shell context
!bash -c "nohup {ollama_path} serve > ollama_serve.log 2>&1 &"

# 3. Robustly check if Ollama server is up and reachable
server_up = False
max_attempts = 15
for i in range(max_attempts):
    try:
        response = requests.get('http://127.0.0.1:11434/api/tags', timeout=10)
        if response.status_code == 200:
            print("Ollama server is up and reachable!")
            server_up = True
            break
    except requests.exceptions.ConnectionError:
        print(f"Attempt {i+1}/{max_attempts}: Ollama server not yet reachable. Retrying in 10 seconds...")
    except requests.exceptions.Timeout:
        print(f"Attempt {i+1}/{max_attempts}: Connection timed out. Retrying in 10 seconds...")
    time.sleep(10)

if not server_up:
    print("Failed to connect to Ollama server after multiple attempts.")
    print("Checking ollama_serve.log for more details:")
    !cat ollama_serve.log
else:
    print("Successfully connected to Ollama server. Now checking for llama3.2 model...")

    # 4. Check if llama3.2 model is available, if not, pull it
    model_available = False
    try:
        response = requests.get('http://127.0.0.1:11434/api/tags', timeout=10)
        if response.status_code == 200:
            models_info = response.json()
            if any(model['name'].startswith('llama3.2') for model in models_info.get('models', [])):
                model_available = True
    except requests.exceptions.RequestException as e:
        print(f"Error checking models: {e}")

    if not model_available:
        print('llama3.2 model not found. Attempting to pull it now...')
        !bash -c "{ollama_path} pull llama3.2 > ollama_pull_log.txt 2>&1" # Use bash -c for pull as well
        print('\nOllama pull log:')
        !cat ollama_pull_log.txt

        # Re-check model availability after pull attempt
        time.sleep(5)
        try:
            response = requests.get('http://127.0.0.1:11434/api/tags', timeout=10)
            if response.status_code == 200:
                models_info = response.json()
                if any(model['name'].startswith('llama3.2') for model in models_info.get('models', [])):
                    model_available = True
        except requests.exceptions.RequestException as e:
            print(f"Error re-checking models after pull: {e}")

    if model_available:
        print('llama3.2 model is available! Proceeding with query.')
        # 5. Run the query and capture its full output
        !bash -c "{ollama_path} run llama3.2 \"GET ME SOME POWERPACKED BREKFAST RECIPIES.\" > ollama_query_log.txt 2>&1" # Use bash -c for run
        print('\nOllama query log:')
        !cat ollama_query_log.txt
    else:
        print('llama3.2 model is still not found after pull attempt. Please review ollama_pull_log.txt for errors.')
```
